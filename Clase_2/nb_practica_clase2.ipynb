## 1. Datasets Clásicos de Benchmarking y Estudio

Para validar algoritmos de manera controlada sin requerir procesos prolongados de limpieza o depuración, la comunidad de ciencia de datos recurre a conjuntos experimentales estandarizados:

| Dataset | Tipo de Problema | Dominio | Descripción |
| :--- | :--- | :--- | :--- |
| **[Iris Dataset](https://archive.ics.uci.edu/ml/datasets/iris)** | Clasificación (Tabular) | Botánica | Clasificación de 3 especies de flores en base a 4 medidas físicas del sépalo y pétalo. Es el clásico "Hello World" en modelado estadístico supervisado. |
| **[MNIST](http://yann.lecun.com/exdb/mnist/)** | Clasificación (Imágenes) | Visión / OCR | 70,000 imágenes en blanco y negro (28x28 px) de dígitos numéricos escritos a mano (0 al 9). El estándar introductorio universal para redes neuronales computacionales. |
| **[California Housing](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.fetch_california_housing.html)** | Regresión (Tabular) | Bienes Raíces | Predicción de valores de viviendas basada en tasas criminales, cercanía al centro, impuestos y zona ocupada (reemplazando a Boston Housing por consideraciones modernas). |
| **[UCI ML Repository](https://archive.ics.uci.edu/)** | Todos los tipos | Multidisciplinario | El mayor archivo académico universitario de datasets de libre acceso para todo tipo de tareas estadístico-matemáticas de Machine Learning. |

```python
# Ejemplo Práctico 1: Cargar datasets canónicos desde scikit-learn y explorarlos con Pandas
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets

# Configuración visual para los gráficos del notebook
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (8, 4)

# 1. Cargar Iris Dataset
iris = datasets.load_iris()
df_iris = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df_iris['especie'] = [iris.target_names[i] for i in iris.target]

print("--- DIMENSIONES IRIS DATASET ---")
print(f"Filas: {df_iris.shape[0]}, Columnas: {df_iris.shape[1]}")
display(df_iris.head(3))

# Gráfica de dispersión de Iris por Especie
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df_iris, x="sepal length (cm)", y="petal length (cm)", hue="especie", palette="Dark2", s=70)
plt.title("Iris Dataset - Largo del Sépalo vs Largo del Pétalo por Especie")
plt.show()
```

## 2. Repositorios de Imágenes y Visión Artificial (Computer Vision)

Esenciales para aquellos proyectos aplicados a robótica autónoma, inspección de calidad industrial mediante cámaras o clasificación en electrónica embebida:

* **[Roboflow Universe](https://universe.roboflow.com/):** Plataforma masiva de la comunidad open-source para visión por computador. Cuenta con cientos de miles de conjuntos de imágenes debidamente etiquetados y listos para exportar a arquitecturas populares como YOLO, Fast-RCNN o TensorFlow Lite.
    * *Ejemplos aplicables al curso:* Detección de fallas en soldadura o componentes sobre placas PCB, conteo de piezas en líneas de ensamblaje o lectura automática de medidores de energía.
* **[MS COCO (Common Objects in Context)](https://cocodataset.org/) & [ImageNet](https://www.image-net.org/):** Colecciones monumentales para detección, segmentación y reconocimiento general en deep learning masivo.
* **[Kaggle (Computer Vision)](https://www.kaggle.com/datasets?tags=13204-Computer+Vision):** Permite filtrar los catálogos específicamente por tags de visión artificial para encontrar desde fotografías satelitales térmicas hasta reconocimiento de piezas defectuosas en fábricas.

```python
# Ejemplo Práctico 2: Representación matricial en Computer Vision (Cómo ve Python una imagen 28x28)
import numpy as np

# Simulamos el formato numérico de una imagen monocroma (matriz 28x28 en escala de grises de 0 a 255)
np.random.seed(42)
matriz_imagen = np.zeros((28, 28))

# Dibujamos un patrón (ejemplo: simulación de una pista conductiva o dígito numérico en inspección óptica)
matriz_imagen[5:23, 12:15] = 240  # Línea vertical brillante
matriz_imagen[12:15, 6:22] = 240  # Línea horizontal (cruz)
matriz_imagen += np.random.normal(10, 5, (28, 28))  # Añadir ruido térmico de sensor

plt.figure(figsize=(5, 4))
plt.imshow(matriz_imagen, cmap='magma', aspect='auto')
plt.colorbar(label='Intensidad del Píxel (0 = Negro, 255 = Blanco)')
plt.title("Matriz de 28x28 píxeles (Visión Artificial / Análisis Óptico)")
plt.grid(False)
plt.show()

print("Forma matricial del tensor:", matriz_imagen.shape)
```

## 3. Repositorios de Datos Climáticos y Meteorológicos

Las series temporales atmosféricas representan una excelente área de estudio en ingeniería para la predicción de consumo en redes, generación solar/eólica o alertas hidrometeorológicas:

* **[Copernicus Climate Data Store (CDS / ECMWF - ERA5)](https://cds.climate.copernicus.eu/):** Infraestructura de la Agencia Espacial Europea que provee reanálisis y pronósticos planetarios históricos hora a hora por longitud y latitud del planeta entero (temperatura, presión, radiación solar, humedad, viento).
* **[NOAA National Centers for Environmental Information](https://www.ncei.noaa.gov/):** Bases de datos históricas federales norteamericanas sobre clima marítimo, atmosférico y radares precipitacionales.
* **[Open-Meteo Historical Weather API](https://open-meteo.com/):** Extraordinaria API gratuita y abierta de archivos climáticos históricos de cualquier punto del mundo al instante. Permite descargar reportes continuados en formato tabular en Python de forma ultraparida y sin llaves complicadas.

```python
# Ejemplo Práctico 3: Descargar y graficar datos meteorológicos reales de Asunción, Paraguay (Open-Meteo REST API)
import requests

# Coordenadas de Asunción, Paraguay (Latitud: -25.26, Longitud: -57.57) - Año de muestra: 2023
url_clima = "https://archive-api.open-meteo.com/v1/archive?latitude=-25.26&longitude=-57.57&start_date=2023-01-01&end_date=2023-12-31&daily=temperature_2m_max,temperature_2m_min,precipitation_sum&timezone=America%2FSao_Paulo"

try:
    respuesta = requests.get(url_clima, timeout=10)
    datos_clima = respuesta.json()
    
    df_asuncion = pd.DataFrame(datos_clima['daily'])
    df_asuncion['time'] = pd.to_datetime(df_asuncion['time'])
    
    print("--- MUESTRA CLIMÁTICA HORARIA ASUNCIÓN (OPEN-METEO) ---")
    display(df_asuncion.head())
    
    # Graficamos la serie temporal del clima
    plt.figure(figsize=(10, 4))
    plt.plot(df_asuncion['time'], df_asuncion['temperature_2m_max'], label='Temp. Máxima (°C)', color='#DC2626', linewidth=1.2)
    plt.plot(df_asuncion['time'], df_asuncion['temperature_2m_min'], label='Temp. Mínima (°C)', color='#2563EB', linewidth=1.2)
    plt.fill_between(df_asuncion['time'], df_asuncion['temperature_2m_max'], df_asuncion['temperature_2m_min'], color='#E2E8F0', alpha=0.5)
    plt.title("Temperatura Diaria en Asunción (Año 2023) - Fuente: Open-Meteo")
    plt.xlabel("Fecha"); plt.ylabel("Temperatura (°C)")
    plt.legend(loc="upper right")
    plt.tight_layout()
    plt.show()

except Exception as e:
    print("Nota: Si no posees conexión a Internet en este momento para la consulta HTTP, aquí se mostraría el gráfico histórico.", e)
```

## 4. Repositorios Demográficos, Sociales y de Salud (Healthcare & Bio-ML)

Útiles tanto para sociometría urbana y proyecciones de conectividad en redes sanitarias/eléctricas como para Electrónica Médica:

### Demografía y Sociedad
* **[World Bank Open Data (Banco Mundial)](https://data.worldbank.org/):** Catálogo masivo con tablas descargables por país sobre infraestructura, electrificación rural e indicadores económicos.
* **[IPUMS International](https://ipums.org/):** El mayor inventario de microdatos censales (hogares e individuos desidentificados) a escala mundial.
* **[Gapminder Open Data](https://www.gapminder.org/data/):** Series longitudinales limpias para estudiar evolución en esperanza de vida e índices globales.

### Enfermedades y Biomedicina (Healthcare & Bio-ML)
* **[PhysioNet (MIMIC-III / MIMIC-IV)](https://physionet.org/):** Base hospitalaria de cuidados intensivos desarrollada por el MIT & Beth Israel Hospital. Dispone de millones de registros desidentificados: señales fisiológicas continuas, monitores ECG, EEG, presión arterial en crudo y laboratorios horarios.
* **[NIH Chest X-ray Dataset](https://www.nih.gov/news-events/news-releases/nih-clinical-center-provides-one-largest-publicly-available-chest-x-ray-datasets-scientific-community):** Más de 100,000 radiografías torácicas anotadas con 14 patologías pulmonares (neumonía, cardiomegalia, efusiones).
* **[COVID-19 Radiography Database (Kaggle)](https://www.kaggle.com/datasets/tawsifurrahman/covid19-radiography-database):** Placas radiográficas divididas en positivos al virus, neumonía viral y pulmón saludable.

```python
# Ejemplo Práctico 4: Análisis de Serie Temporal en Biomedicina (Onda ECG simulada tipo PhysioNet)
import numpy as np

# Simulamos 5 segundos de monitoreo de electrocardiograma (ECG) muestreado a 100 Hz
fs = 100  # Frecuencia de muestreo (Hz)
tiempo_ecg = np.linspace(0, 5, 5 * fs)
onda_base = 0.05 * np.sin(2 * np.pi * 1.0 * tiempo_ecg) + np.random.normal(0, 0.01, len(tiempo_ecg))

# Inyección periódica del complejo QRS (latidos cardiacos cada ~0.8 segundos = 75 BPM)
latidos_segundos = np.arange(0.5, 4.8, 0.8)
for latido in latidos_segundos:
    idx = int(latido * fs)
    if idx < len(onda_base) - 5:
        onda_base[idx - 2] = -0.15  # Onda Q
        onda_base[idx]     = 1.40   # Pico R
        onda_base[idx + 2] = -0.25  # Onda S
        onda_base[idx + 8] = 0.20   # Onda T

df_medico = pd.DataFrame({"tiempo_seg": tiempo_ecg, "senal_ecg_mv": onda_base})

plt.figure(figsize=(10, 3.2))
plt.plot(df_medico['tiempo_seg'], df_medico['senal_ecg_mv'], color='#059669', linewidth=1.5)
plt.title("PhysioNet MIMIC-IV Simulación: Señal ECG en Unidad de Cuidados Intensivos (75 BPM)")
plt.xlabel("Tiempo (segundos)"); plt.ylabel("Amplitud Eléctrica (mV)")
plt.xlim(0, 5)
plt.tight_layout()
plt.show()
```

## 5. Caso de Estudio e Integración Regional: Datos Abiertos de Paraguay (PY)

Para desarrollar proyectos de ingeniería enfocados en solventar problemáticas energéticas, climáticas o sanitarias en territorio paraguayo, contamos con excelentes fuentes institucionales interoperables:

1. **[Portal Oficial de Datos Abiertos (`www.datos.gov.py`)](https://www.datos.gov.py/):** Centraliza compras del estado (DNCP), becas educativas, inversiones en infraestructura e índices económicos en formatos JSON o CSV.
2. **[INE (Instituto Nacional de Estadística de Paraguay)](https://www.ine.gov.py/):** Acceso a microdatos inestimables de la **Encuesta Permanente de Hogares Continua (EPHC)** y del **Censo Nacional 2022** para estimaciones demográficas regionales por departamento.
3. **[MSPBS (Dirección de Vigilancia de la Salud)](https://dgvs.mspbs.gov.py/):** Boletines e informes históricos sobre brotes locales de **Arbovirosis (Dengue, Chikungunya, Zika)**. Excelente para modelado predictivo espacio-temporal asociado a lluvias y temperatura.
4. **[Sector Eléctrico (ANDE / Itaipu / Yacyretá / VME)](https://www.ande.gov.py/):** Registros de demanda y curva de carga horaria en el **Sistema Interconectado Nacional (SIN)**, un problema rey de serie temporal para electrónica de potencia y redes inteligentes (Smart Grid).
5. **[Dirección de Meteorología e Hidrología (DMH / DINAC)](https://www.meteorologia.gov.py/):** Registros pluviométricos e hidrométricos de los ríos Paraguay y Paraná para pronóstico y alertas de estiaje o crecidas.

```python
# Ejemplo Práctico 5 (Caso de Estudio): Correlación de Demanda Eléctrica del SIN (ANDE) vs Temperatura en Asunción
import numpy as np
import seaborn as sns

# Generamos una tabla realista simulada de 150 observaciones horarias de verano en Asunción
np.random.seed(88)
temp_asuncion = np.random.uniform(19, 43, 150)  # Temperaturas desde mañana fresca hasta siesta torrada
# La demanda se dispara exponencialmente cuando la temperatura supera los 28°C debido al uso de aires acondicionados
demanda_sin_mw = 2100 + 12 * temp_asuncion + 2.5 * np.maximum(0, temp_asuncion - 27)**2.3 + np.random.normal(0, 45, 150)

df_py = pd.DataFrame({
    "temperatura_ambiente_c": np.round(temp_asuncion, 1),
    "demanda_activa_mw": np.round(demanda_sin_mw, 1)
})

print("--- MUESTRA DE DATOS DEL SISTEMA INTERCONECTADO NACIONAL (PY) ---")
display(df_py.describe())

# Análisis de correlación visual e inspección del efecto del aire acondicionado
plt.figure(figsize=(8, 4.5))
sns.regplot(
    data=df_py, x="temperatura_ambiente_c", y="demanda_activa_mw", order=2,
    scatter_kws={'alpha': 0.7, 'color': '#D97706', 's': 45},
    line_kws={'color': '#1E3A8A', 'linewidth': 2.5, 'label': 'Ajuste Polinómico 2° Grado'}
)

plt.axvline(28, color='red', linestyle='--', alpha=0.6, label='Umbral Crítico Aire Acondicionado (28°C)')
plt.title("Paraguay (ANDE) - Correlación No Lineal: Demanda Activa del SIN vs Temperatura")
plt.xlabel("Temperatura Ambiente en Asunción (°C)")
plt.ylabel("Demanda Activa Eléctrica (MW)")
plt.legend(loc="upper left")
plt.tight_layout()
plt.show()
```
