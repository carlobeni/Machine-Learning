## 2. Programación Orientada a Objetos (POO)

La Programación Orientada a Objetos es un paradigma que permite modelar sistemas complejos mediante **abstracciones**, agrupando datos y comportamientos en entidades llamadas **objetos**.

Desde el punto de vista de la ingeniería, la POO permite:

- Agrupar datos y procedimientos relacionados
- Reducir la complejidad mediante diseño modular
- Facilitar el mantenimiento y la extensión del software
- Reutilizar código de manera eficiente
- Implementar sistemas escalables y jerárquicos


### 2.1 ¿Qué es una clase y qué es un objeto?

Una **clase** es una **abstracción** (generalización, plantilla o template) que define un conjunto de:

- **Atributos**: representación interna de los datos
- **Métodos**: procedimientos que definen el comportamiento del objeto

Un **objeto** es una **instancia concreta** de una clase.

#### Ejemplo conceptual 1 (robótica)

La clase `Robot` puede representar de forma abstracta distintos sistemas físicos:

- Drones
- Robots industriales
- Robots aspiradores

Todos ellos comparten atributos como:
- Motores
- Sensores
- Controladores

Y comportamientos como:
- Desplazamiento
- Orientación
- Lectura de sensores

#### Ejemplo conceptual 2 (interfaces gráficas)

Una ventana de Windows es una abstracción común a:
- Spotify
- Gestor de archivos
- Google Chrome
- Visual Studio Code

Todas poseen atributos como alto y ancho, y responden de forma similar a eventos como cerrar o redimensionar.


### 2.2 ¿Cómo crear una clase?

Python permite definir **nuevos tipos de datos** mediante la palabra clave `class`.  
Las clases encapsulan la representación interna y exponen una **interfaz** para interactuar con los objetos, ocultando los detalles de implementación.

En proyectos simples, los tipos nativos suelen ser suficientes.  
Sin embargo, a medida que un sistema crece en complejidad, es común integrar y desarrollar clases que representen:

- Arrays numéricos (`numpy.ndarray`)
- Nodos, sensores y actuadores en **ROS2**
- Modelos de visión artificial e inteligencia artificial de **YOLO**
- Elementos gráficos como los de **PyQt6**

Las clases permiten unir estos elementos de distintas áreas en estructuras coherentes y reutilizables, aplicando el principio **divide y vencerás**.

#### Formato básico general de una clase



```python
class NameClass:
    def __init__(self, param_value1, param_value2):
        self.__attribute1 = param_value1
        self.__attribute2 = param_value2
        # Otros atributos

    def method1(self, param1, param2):
        # Código del método
        pass

    # Otros métodos
```

#### Ejemplo: Clase `Robot`



```python
class Robot:
    def __init__(self, num_motors, robot_type):
        self.__num_motors = num_motors
        self.__robot_type = robot_type
        self.__is_active = False

    def start(self):
        self.__is_active = True
        print("Robot activado")

    def stop(self):
        self.__is_active = False
        print("Robot detenido")

    def __str__(self):
        return f"Robot tipo {self.__robot_type} con {self.__num_motors} motores"
```

### 2.3. Buenas prácticas de programación orientada a objetos

Para escribir código profesional, debemos seguir principios de **ocultamiento de información** y **encapsulamiento**.

#### A. Atributos Privados
En Python, para indicar que un atributo es privado y que no debe ser accedido directamente desde fuera de la clase, se antepone un doble guion bajo (`__`). Esto activa el *name mangling* para proteger la integridad del dato.


```python
class Robot:
    def __init__(self, num_motors, robot_type):
        self.__num_motors = num_motors
        self.__robot_type = robot_type
        self.__is_active = False
        self.__is_moving = False
```

#### B. Decoradores y Lanzamiento de Excepciones
En lugar de acceder a los atributos directamente, usamos el decorador `@property`. Esto permite validar datos antes de asignarlos (por ejemplo, impedir velocidades negativas en un motor). Ademas en lugar de solo imprimir un mensaje de error, lanzamos una excepción para que el programa pueda capturar y manejar el error de forma robusta


```python
class Robot:
    def __init__(self, num_motors, robot_type):
        self.__num_motors = num_motors
        self.__robot_type = robot_type  
        self.__is_active = False
        self.__is_moving = False
        
    @property
    def is_active(self):
        return self.__is_active
    
    @is_active.setter
    def is_active(self, value):
        if not isinstance(value, bool):
            raise ValueError("El estado 'is_active' debe ser de tipo Booleano.")    
            
        self.__is_active = value    
    
    @property
    def is_moving(self):
        return self.__is_moving
    
    @is_moving.setter
    def is_moving(self, value):
        if not isinstance(value, bool):
            raise ValueError("El estado 'is_moving' debe ser de tipo Booleano.")    
            
        self.__is_moving = value
```

#### C. Sobrecarga de Operadores (Dunder Methods)
Podemos definir cómo se comportan nuestros objetos con operadores matemáticos o funciones integradas (como `+`, `-`, `==` o `print`) usando métodos especiales como `__add__`, `__eq__` o `__str__`.

Tabla de funciones especiales para operadores:
| Operador | Método especial |
| --- | --- |
| `+` | `__add__` |
| `-` | `__sub__` |
| `*` | `__mul__` |
| `/` | `__truediv__` |
| `%` | `__mod__` |
| `**` | `__pow__` |
| `//` | `__floordiv__` |
| `<<` | `__lshift__` |
| `>>` | `__rshift__` |
| `&` | `__and__` |
| `\|` | `__or__` |
| `^` | `__xor__` |
| `~` | `__inv__` |
| `==` | `__eq__` |
| `!=` | `__ne__` |
| `<` | `__lt__` |
| `<=` | `__le__` |
| `>` | `__gt__` |
| `>=` | `__ge__` |
| `is` | `__is__` |
| `is not` | `__is_not__` |



```python
class Robot:
    def __init__(self, num_motors, robot_type):
        self.__num_motors = num_motors
        self.__robot_type = robot_type
        self.__is_active = False
        self.__is_moving = False

    def __str__(self):
        return f"Robot {self.__robot_type} | Motores: {self.__num_motors}"

    def __eq__(self, other):
        return self.__robot_type == other.__robot_type

    def __add__(self, other):
        return Robot(self.__num_motors + other.__num_motors, self.__robot_type)
```

Otras buenas prácticas destacadas son:
- Usar nombres descriptivos para los objetos y clases evitando usar nombres genericos o ambiguos
- Los nombres de las clases deben comenzar con una letra mayúscula, y objetos, los atributos y métodos deben comenzar con una letra minúscula

### 2.5. Herencia
La herencia permite crear nuevas clases a partir de clases existentes, reutilizando atributos y métodos previamente definidos.
- Clase padre (superclase): clase base
- Clase hija (subclase): clase derivada que hereda y puede extender o redefinir comportamientos

La herencia es clave para construir jerarquías de clases y sistemas escalables.

#### Formato básico de herencia


```python
class NameClass(NameClass2):
    def __init__(self, param_value1, param_value2):
        super().__init__(param_value1, param_value2)
        self.__attribute1 = param_value1
        self.__attribute2 = param_value2
        # Otros atributos

    def method1(self, param1, param2):
        # Código del método
        pass

    # Otros métodos
```

#### Ejemplo: Herencia de clase `Robot`


```python
class Dron(Robot):
    def __init__(self, num_motors, weight):
        super().__init__(num_motors, "dron")
        self.__weight = weight

    def fly(self):
        print("Volando")

    def __str__(self):
        return f"Dron de {self.__weight} kg con {self.__num_motors} motores"

my_dron = Dron(4, 1.5)
my_dron.start()
my_dron.fly()
```

## 3. Introducción a NumPy

NumPy (Numerical Python) es una librería fundamental para la computación científica en Python. Proporciona soporte para arreglos y matrices multidimensionales, junto con una colección de funciones matemáticas de alto nivel para operar con estos arreglos.

### 3.1. Creación y operaciones básicas


```python
import numpy as np

# Crear un arreglo de 1D y 2D
arr_1d = np.array([1, 2, 3, 4, 5])
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("Arreglo 1D:\n", arr_1d)
print("Arreglo 2D:\n", arr_2d)

# Operaciones matemáticas vectorizadas
print("\nArreglo al cuadrado:\n", arr_1d ** 2)
print("Suma del arreglo 1D:", np.sum(arr_1d))
```

### 3.2. Indexación, Slicing y Extracción de datos
NumPy facilita la extracción de porciones específicas de datos de un arreglo (slicing) o la selección mediante índices.


```python
# Acceder a un elemento específico (fila, columna)
print("Elemento en fila 1, columna 2:", arr_2d[1, 2])

# Extraer una fila completa
print("Fila 0 completa:", arr_2d[0, :])

# Extraer una columna completa
print("Columna 1 completa:", arr_2d[:, 1])

# Slicing en 1D (del índice 1 al 3)
print("Sub-arreglo 1D:", arr_1d[1:4])
```

### 3.3. Filtrado Booleano
Una técnica ampliamente utilizada en ciencia de datos es usar condiciones para filtrar arreglos, lo que devuelve solo los elementos que cumplen la condición.


```python
edades = np.array([15, 22, 18, 30, 12, 40])

# Obtener una máscara booleana
mayores_de_edad = edades >= 18
print("Máscara booleana:", mayores_de_edad)

# Filtrar aplicando la máscara
print("Edades filtradas (>= 18):", edades[mayores_de_edad])
```

## 4. Introducción a Pandas

Pandas es una librería de software escrita como extensión de NumPy para manipulación y análisis de datos. Ofrece estructuras de datos y operaciones para manipular tablas numéricas y series temporales mediante la clase `DataFrame`.

### 4.1. Creación y exploración básica


```python
import pandas as pd
import numpy as np

# Crear un DataFrame a partir de un diccionario
data = {
    'Nombre': ['Ana', 'Luis', 'Carlos', 'Marta', 'Pedro', 'Sofia'],
    'Edad': [28, 34, 29, 42, np.nan, 25],  # Un valor faltante (NaN)
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla', 'Madrid', 'Bilbao'],
    'Salario': [2500, 3200, 2800, 4100, 3000, np.nan]
}
df = pd.DataFrame(data)

# Mostrar las primeras filas del DataFrame
print("Primeras filas:\n", df.head())

# Estadísticas básicas
print("\nEstadísticas del dataset:\n", df.describe())
```

### 4.2. Extracción de datos (Loc e Iloc)
Extraer columnas, filas o celdas específicas es una de las tareas más comunes en análisis de datos.


```python
# Extraer una columna
print("Edades:\n", df['Edad'])

# Extraer múltiples columnas
print("\nNombres y Ciudades:\n", df[['Nombre', 'Ciudad']])

# iloc: Seleccionar por índice numérico (Filas 0 y 1)
print("\nPrimeras 2 filas (iloc):\n", df.iloc[0:2])

# loc: Seleccionar por etiquetas/condiciones de índice
print("\nFila del índice 2 (loc):\n", df.loc[2])
```

### 4.3. Manejo de valores nulos (NaNs)
Los datos reales a menudo están incompletos. Pandas tiene herramientas para encontrar y manejar valores faltantes (`NaN`).


```python
# Identificar qué celdas son nulas
print("Valores nulos por columna:\n", df.isna().sum())

# Estrategia 1: Eliminar las filas que contengan algún NaN
df_limpio = df.dropna()
print("\nDataFrame sin NaNs:\n", df_limpio)

# Estrategia 2: Rellenar los valores nulos con otro valor (ej. la media o un valor fijo)
df_relleno = df.copy()
df_relleno['Edad'] = df_relleno['Edad'].fillna(df['Edad'].mean())
df_relleno['Salario'] = df_relleno['Salario'].fillna(0)
print("\nDataFrame con NaNs rellenados:\n", df_relleno)
```

### 4.4. Filtrado de datos avanzado
Podemos combinar condiciones lógicas para filtrar nuestro `DataFrame`.


```python
# Filtrar personas mayores de 30 años que sean de Barcelona
filtro = (df['Edad'] > 30) & (df['Ciudad'] == 'Barcelona')

# Aplicar el filtro
print("Resultado del filtro:\n", df[filtro])
```

## 5. Introducción a Matplotlib

Matplotlib es la librería de generación de gráficos más utilizada en Python, permitiendo crear visualizaciones estáticas, animadas e interactivas de alta calidad. El módulo `pyplot` es su componente principal, que provee una interfaz similar a la de MATLAB.

### 5.1. Gráficos de línea básicos
Los gráficos de línea son ideales para representar series continuas o tendencias.


```python
import matplotlib.pyplot as plt

# Datos de ejemplo
x = np.linspace(0, 10, 100) # 100 puntos entre 0 y 10
y1 = np.sin(x)
y2 = np.cos(x)

# Crear el gráfico
plt.figure(figsize=(8, 4)) # Tamaño de la figura (ancho, alto)
plt.plot(x, y1, label='Seno(x)', color='blue', linestyle='-')
plt.plot(x, y2, label='Coseno(x)', color='red', linestyle='--')

# Configurar etiquetas y título
plt.title('Funciones Trigonométricas')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.legend() # Mostrar las leyendas (labels)
plt.grid(True) # Activar grilla

# Mostrar el gráfico (En un notebook, aparece debajo de la celda)
plt.show()
```

### 5.2. Gráficos de Dispersión (Scatter) y Barras
Matplotlib admite diversos tipos de gráficos que son clave para la exploración de datos.


```python
# Datos categóricos y de dispersión
categorias = ['A', 'B', 'C', 'D']
valores = [15, 30, 45, 20]

puntos_x = np.random.rand(50) * 10
puntos_y = puntos_x * 2 + np.random.randn(50) * 2

# Crear un layout con 2 subgráficos (1 fila, 2 columnas)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Gráfico de barras en el primer subgráfico (ax1)
ax1.bar(categorias, valores, color=['orange', 'green', 'purple', 'cyan'])
ax1.set_title('Gráfico de Barras')
ax1.set_ylabel('Cantidades')

# Gráfico de dispersión en el segundo subgráfico (ax2)
ax2.scatter(puntos_x, puntos_y, color='darkblue', alpha=0.6)
ax2.set_title('Gráfico de Dispersión (Scatter)')
ax2.set_xlabel('Variable X')
ax2.set_ylabel('Variable Y')

plt.tight_layout() # Ajustar espaciado
plt.show()
```
