## 1. Definición Formal de Machine Learning

### 1.1. Definición de Tom Mitchell (1997)

> *"Un programa de computadora **aprende** de la experiencia **E** con respecto a una tarea **T** y una medida de rendimiento **P**, si su rendimiento en **T**, medido por **P**, mejora con **E**."*

### 1.2. Componentes de la definición

| **Símbolo** | **Significado** | **Ejemplo en Ingeniería** |
|-------------|-----------------|---------------------------|
| $T$ | Tarea a realizar | Clasificar señales ECG |
| $P$ | Medida de rendimiento | Precisión (Accuracy) |
| $E$ | Experiencia (datos) | 10,000 registros etiquetados |

### 1.3. Formalización Matemática

El aprendizaje puede expresarse como un problema de optimización:

$$
f^* = \arg\min_{f \in \mathcal{H}} \frac{1}{N} \sum_{i=1}^{N} L(y_i, f(x_i)) + \lambda \Omega(f)
$$

Donde:
- $\mathcal{H}$: espacio de hipótesis (modelos posibles)
- $L$: función de pérdida (mide el error)
- $\Omega(f)$: término de regularización (penaliza complejidad)
- $\lambda$: hiperparámetro que controla la regularización

## 2. El Ciclo de Vida de un Proyecto de ML

### 2.1. CRISP-DM (Cross-Industry Standard Process for Data Mining)

```
┌─────────────────────────────────────────────────────────────┐
│                    CRISP-DM                                │
├─────────────────────────────────────────────────────────────┤
│                                                             │
│  1. Entendimiento del negocio/problema                      │
│         ↓                                                   │
│  2. Obtención y exploración de datos (EDA)                 │
│         ↓                                                   │
│  3. Preprocesamiento y feature engineering                 │
│         ↓                                                   │
│  4. Modelado y validación                                  │
│         ↓                                                   │
│  5. Evaluación y ajuste de hiperparámetros                 │
│         ↓                                                   │
│  6. Despliegue y monitoreo                                 │
│                                                             │
└─────────────────────────────────────────────────────────────┘
```

**Importante:** El flujo es iterativo, no lineal. Se puede volver a etapas anteriores.

## 3. Tipos de Aprendizaje

### 3.1. Aprendizaje Supervisado (Supervised Learning)

#### 3.1.1. Definición
El algoritmo aprende a mapear entradas $x$ a salidas $y$ utilizando un conjunto de datos etiquetados.

#### 3.1.2. Formulación Matemática

Dado un conjunto de datos de entrenamiento:

$$
\mathcal{D} = \{(x_i, y_i)\}_{i=1}^{N} \quad \text{donde} \quad x_i \in \mathcal{X}, \; y_i \in \mathcal{Y}
$$

El objetivo es encontrar una función $f: \mathcal{X} \to \mathcal{Y}$ que minimice:

$$
\mathcal{L}(f) = \frac{1}{N} \sum_{i=1}^{N} L(y_i, f(x_i)) + \lambda \Omega(f)
$$

#### 3.1.3. Sub-tipos:

| **Tipo** | **Variable de salida** | **Ejemplo** |
|----------|----------------------|-------------|
| **Regresión** | Continua ($y \in \mathbb{R}$) | Predecir precio de vivienda |
| **Clasificación** | Discreta/Categórica ($y \in \{1, \dots, K\}$) | Identificar dígitos manuscritos |

#### 3.1.4. Algoritmos comunes
- Regresión Lineal
- Regresión Logística
- Máquinas de Soporte Vectorial (SVM)
- Árboles de Decisión
- Redes Neuronales
- k-Nearest Neighbors (k-NN)
- Random Forest
- Gradient Boosting

#### 3.1.5. Regresión Lineal

**Modelo:**

$$
\hat{y} = w^T x + b = \sum_{j=1}^{d} w_j x_j + b
$$

**Función de pérdida (MSE - Error Cuadrático Medio):**

$$
J(w, b) = \frac{1}{2N} \sum_{i=1}^{N} \left( y_i - \hat{y}_i \right)^2
$$

**Solución analítica (mínimos cuadrados):**

$$
\hat{w} = (X^T X)^{-1} X^T y
$$

**Gradiente para optimización (SGD):**

$$
\frac{\partial J}{\partial w_j} = \frac{1}{N} \sum_{i=1}^{N} (\hat{y}_i - y_i) x_{ij}
$$

#### 3.1.6. Clasificación Binaria - Regresión Logística

**Función Sigmoide:**

$$
\sigma(z) = \frac{1}{1 + e^{-z}} \quad \Longrightarrow \quad \hat{y} = \sigma(w^T x + b)
$$

**Interpretación probabilística:**

$$
P(y=1|x) = \hat{y}, \quad P(y=0|x) = 1 - \hat{y}
$$

**Función de pérdida (Cross-Entropy / Log Loss):**

$$
J(w, b) = -\frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]
$$

#### 3.1.7. Clasificación Multiclase - Softmax

**Función Softmax:**

$$
\hat{y}_k = \frac{e^{z_k}}{\sum_{j=1}^{K} e^{z_j}} \quad \text{para } k = 1, \dots, K
$$

**Interpretación probabilística:**

$$
P(y = k | x) = \hat{y}_k
$$

**Función de pérdida (Cross-Entropy multiclase):**

$$
J(w, b) = -\sum_{i=1}^{N} \sum_{k=1}^{K} y_{ik} \log(\hat{y}_{ik})
$$

### 3.2. Aprendizaje No Supervisado (Unsupervised Learning)

#### 3.2.1. Definición
El algoritmo encuentra estructuras ocultas o patrones en datos no etiquetados.

#### 3.2.2. Formulación Matemática

Dado un conjunto de datos de entrenamiento:

$$
\mathcal{D} = \{x_i\}_{i=1}^{N} \quad \text{donde} \quad x_i \in \mathcal{X}
$$

El objetivo es encontrar una estructura $g: \mathcal{X} \to \mathcal{Z}$ que revele patrones:

$$
g^* = \arg\min_{g \in \mathcal{G}} \sum_{i=1}^{N} d(x_i, g(x_i))
$$

#### 3.2.3. Sub-tipos:

| **Tipo** | **Objetivo** | **Ejemplo** |
|----------|-------------|-------------|
| **Clustering** | Agrupar datos similares | Segmentación de clientes |
| **Reducción de dimensionalidad** | Proyectar a espacio de menor dimensión | Visualización de datos |
| **Detección de anomalías** | Encontrar puntos atípicos | Detección de fraudes |

#### 3.2.4. Algoritmos comunes
- K-Means
- Análisis de Componentes Principales (PCA)
- t-SNE
- DBSCAN
- Autoencoders
- Hierarchical Clustering
- Gaussian Mixture Models (GMM)

#### 3.2.5. K-Means Clustering

**Objetivo:** Minimizar la inercia intra-cluster

$$
J = \sum_{k=1}^{K} \sum_{i \in C_k} \|x_i - \mu_k\|^2
$$

**Algoritmo:**

1. Inicializar $K$ centroides $\mu_k$ aleatoriamente
2. Asignar cada punto al centroide más cercano:

$$
C_k = \{x_i : \|x_i - \mu_k\|^2 \leq \|x_i - \mu_j\|^2 \; \forall j, 1 \leq j \leq K\}
$$

3. Recalcular centroides como media del cluster:

$$
\mu_k = \frac{1}{|C_k|} \sum_{x_i \in C_k} x_i
$$

4. Repetir pasos 2-3 hasta convergencia

**Método del Codo (Elbow Method) para elegir K:**

$$
WSS(K) = \sum_{k=1}^{K} \sum_{x_i \in C_k} \|x_i - \mu_k\|^2
$$

Se elige $K$ en el punto donde la curva de WSS comienza a aplanarse.

**Coeficiente de Silueta (Silhouette Score):**

$$
s(i) = \frac{b(i) - a(i)}{\max\{a(i), b(i)\}}
$$

Donde:
- $a(i)$: distancia media al resto de puntos del mismo cluster
- $b(i)$: distancia media al cluster más cercano

#### 3.2.6. Análisis de Componentes Principales (PCA)

**Objetivo:** Proyectar datos de alta dimensión a un espacio de menor dimensión preservando la máxima varianza.

**Pasos:**

1. Estandarizar los datos (media 0, desviación 1):

$$
z_{ij} = \frac{x_{ij} - \mu_j}{\sigma_j}
$$

2. Calcular la matriz de covarianza:

$$
\Sigma = \frac{1}{N-1} Z^T Z
$$

3. Calcular autovalores $\lambda$ y autovectores $v$:

$$
\Sigma v = \lambda v
$$

4. Seleccionar los $k$ autovectores con mayores autovalores

5. Proyectar los datos:

$$
Z_{\text{proj}} = Z W_k
$$

Donde $W_k$ es la matriz de los $k$ autovectores principales.

**Varianza explicada:**

$$
\text{VarExplicada}(k) = \frac{\sum_{j=1}^{k} \lambda_j}{\sum_{j=1}^{d} \lambda_j}
$$

### 3.3. Aprendizaje por Refuerzo (Reinforcement Learning)

#### 3.3.1. Definición
Un agente aprende a tomar decisiones secuenciales interactuando con un entorno, recibiendo recompensas o castigos.

#### 3.3.2. Formulación Matemática

**Proceso de Decisión de Markov (MDP):**

Un MDP se define por la tupla $(\mathcal{S}, \mathcal{A}, P, R, \gamma)$ donde:

- $\mathcal{S}$: conjunto de estados
- $\mathcal{A}$: conjunto de acciones
- $P(s'|s, a)$: probabilidad de transición al estado $s'$
- $R(s, a)$: recompensa inmediata
- $\gamma \in [0, 1]$: factor de descuento

**Retorno acumulado:**

$$
G_t = \sum_{k=0}^{\infty} \gamma^k R_{t+k+1} = R_{t+1} + \gamma R_{t+2} + \gamma^2 R_{t+3} + \cdots
$$

**Función de valor (Value Function):**

$$
V^\pi(s) = \mathbb{E}_\pi \left[ G_t \,|\, S_t = s \right]
$$

**Función de valor acción (Q-Function):**

$$
Q^\pi(s, a) = \mathbb{E}_\pi \left[ G_t \,|\, S_t = s, A_t = a \right]
$$

**Ecuación de Bellman para $V^\pi$:**

$$
V^\pi(s) = \sum_{a \in \mathcal{A}} \pi(a|s) \sum_{s' \in \mathcal{S}} P(s'|s, a) \left[ R(s, a) + \gamma V^\pi(s') \right]
$$

**Ecuación de Bellman para $Q^\pi$:**

$$
Q^\pi(s, a) = \sum_{s' \in \mathcal{S}} P(s'|s, a) \left[ R(s, a) + \gamma \sum_{a' \in \mathcal{A}} \pi(a'|s') Q^\pi(s', a') \right]
$$

**Ecuación de Bellman Óptima:**

$$
V^*(s) = \max_{a \in \mathcal{A}} \sum_{s' \in \mathcal{S}} P(s'|s, a) \left[ R(s, a) + \gamma V^*(s') \right]
$$

$$
Q^*(s, a) = \sum_{s' \in \mathcal{S}} P(s'|s, a) \left[ R(s, a) + \gamma \max_{a' \in \mathcal{A}} Q^*(s', a') \right]
$$

**Q-Learning (actualización off-policy):**

$$
Q(s, a) \leftarrow Q(s, a) + \alpha \left[ R(s, a) + \gamma \max_{a'} Q(s', a') - Q(s, a) \right]
$$

**SARSA (actualización on-policy):**

$$
Q(s, a) \leftarrow Q(s, a) + \alpha \left[ R(s, a) + \gamma Q(s', a') - Q(s, a) \right]
$$

**Estrategia $\epsilon$-greedy para exploración:**

$$
\pi(a|s) = \begin{cases}
1 - \epsilon + \frac{\epsilon}{|\mathcal{A}|} & \text{si } a = \arg\max_{a'} Q(s, a') \\
\frac{\epsilon}{|\mathcal{A}|} & \text{en otro caso}
\end{cases}
$$

#### 3.3.3. Algoritmos comunes
- Q-Learning (Tabular)
- SARSA
- Deep Q-Networks (DQN)
- Proximal Policy Optimization (PPO)
- Actor-Critic Methods (A2C, A3C)
- Trust Region Policy Optimization (TRPO)
- Soft Actor-Critic (SAC)

### Ejercicio 3: K-Means a Mano

Dados los siguientes puntos 2D: (1,1), (1,2), (2,1), (8,8), (8,9), (9,8). Realiza K-Means con K=2.

**Paso 1:** Inicializar centroides aleatorios: $\mu_1 = (1,1)$, $\mu_2 = (9,9)$

**Paso 2:** Asignar cada punto al centroide más cercano:
- (1,1) → C1, (1,2) → C1, (2,1) → C1
- (8,8) → C2, (8,9) → C2, (9,8) → C2

**Paso 3:** Recalcular centroides:
- $\mu_1 = (\frac{1+1+2}{3}, \frac{1+2+1}{3}) = (\frac{4}{3}, \frac{4}{3}) \approx (1.33, 1.33)$
- $\mu_2 = (\frac{8+8+9}{3}, \frac{8+9+8}{3}) = (\frac{25}{3}, \frac{25}{3}) \approx (8.33, 8.33)$

**Paso 4:** Repetir hasta convergencia.

El algoritmo converge rápidamente a dos clusters claramente separados.
